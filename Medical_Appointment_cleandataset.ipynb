**Step 1:**

Load the Dataset. I imported the pandas library which is used for data manipulation in Python.

Then I loaded the CSV file into a dataframe and previewed the first 5 rows to understand the structure of the data.

```python
import pandas as pd

df = pd.read_csv('/content/KaggleV2-May-2016.csv')
df.head()
```

**Step 2:**

Explore the Dataset. I used the info() function to check the number of rows and columns, the data type of each column, and whether any column contains missing values.

This helps identify what cleaning is needed before analysis."

```python
df.info()
```

**Step 3:**

Check Column Names. I printed all column names to check for inconsistencies such as mixed capitals, spaces, or special characters that need to be standardized before analysis.

```python
print(df.columns.tolist())
```

**Step 4:**

Standardize Column Names. I cleaned all column names by converting them to lowercase and replacing dashes and spaces with underscores.

This makes the columns consistent and easier to work with in both Python and SQL.

```python
df.columns = df.columns.str.strip().str.lower().str.replace('-', '_').str.replace(' ', '_')
print(df.columns.tolist())
```

**Step 5:**

Check Age Column for Errors. I checked the minimum and maximum age values to identify any unrealistic entries that could affect the analysis.

```python
print(df['age'].min())
print(df['age'].max())
```

**Step 6:**

Remove Invalid Age Values. The minimum age was -1 which is not possible in real life.

I removed that row to ensure the dataset only contains valid age values.

```python
df = df[df['age'] >= 0]
print(df.shape)
```

**Step 7:**

Check Categorical Columns. I checked the unique values in the gender and no_show columns to make sure they only contain expected values and no typos or inconsistencies.

```python
print(df['gender'].unique())
print(df['no_show'].unique())
```

**Step 8:**

Convert Date Columns. The scheduled day and appointment day columns were stored as plain text.

I converted them to proper datetime format so Python can recognize and work with them as dates.

```python
df['scheduledday'] = pd.to_datetime(df['scheduledday'])
df['appointmentday'] = pd.to_datetime(df['appointmentday'])
df.dtypes
```

**Step 9:**

Identify Invalid Date Records. I checked for rows where the appointment date was earlier than the scheduled date, which is logically impossible.

This helps us understand how many records need to be removed.

```python
print(df[df['appointmentday'] < df['scheduledday']].shape)
```

**Step 10:**

Explore Neighbourhood Column. I checked how many unique neighbourhoods exist in the dataset and printed them all to spot any obvious typos, duplicates, or inconsistencies.

```python
print(df['neighbourhood'].nunique())
print(df['neighbourhood'].unique())
```

**Step 11:**

Remove Invalid Date Records. I removed all rows where the appointment date was before the scheduled date as these records are not logically valid and could affect the accuracy of our analysis.

```python
df = df[df['appointmentday'] >= df['scheduledday']]
print(df.shape)
```

**Step 12:**

Verify Date Format. I previewed both date columns to confirm they are now stored as clean date values without any time component attached.

```python
print(df['scheduledday'].head(10))
print(df['appointmentday'].head(10))
```

**Step 13:**

Save Cleaned Dataset. I saved the fully cleaned dataset as a new CSV file.

This keeps the original data untouched while giving us a clean version ready for SQL analysis.

```python
df.to_csv('noshowappointments_cleaned.csv', index=False)
print("File saved successfully!")
```
